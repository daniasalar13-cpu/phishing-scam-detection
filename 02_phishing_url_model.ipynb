```python
from ucimlrepo import fetch_ucirepo

# Load the official UCI PhiUSIIL phishing URL dataset
phiusiil = fetch_ucirepo(id=967)

# Separate features and target
X = phiusiil.data.features
y = phiusiil.data.targets

print("Dataset loaded successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Inspect the dataset
print("Feature columns:")
print(X.columns.tolist())

print("\nFirst 5 rows:")
display(X.head())

print("\nTarget values:")
print(y.head())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
print("Data types:")
print(X.dtypes)

print("\nMissing values:")
print(X.isnull().sum())

print("\nDuplicate rows:")
print(X.duplicated().sum())
```

```python
# Separate text and numerical features

text_columns = X.select_dtypes(include=["object"]).columns.tolist()
numeric_columns = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Text columns:")
for column in text_columns:
    print("•", column)

print("\nNumber of text columns:", len(text_columns))

print("\nNumber of numerical columns:", len(numeric_columns))

print("\nNumerical columns:")
for column in numeric_columns:
    print("•", column)
```

```python
# Check numerical features for missing and infinite values

print("Missing numerical values:")
print(X[numeric_columns].isnull().sum().sum())

print("\nInfinite numerical values:")
print(np.isinf(X[numeric_columns].to_numpy()).sum())

print("\nBasic statistics:")
display(X[numeric_columns].describe().T)
```

```python
from sklearn.model_selection import train_test_split

# Use the 50 numerical features for the first phishing model
X_numeric = X[numeric_columns].copy()

# Convert target to a 1-dimensional array
y_phishing = y["label"]

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X_numeric,
    y_phishing,
    test_size=0.20,
    random_state=42,
    stratify=y_phishing
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create the phishing detection model
phishing_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

# Train the model
phishing_model.fit(X_train, y_train)

print("Phishing model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Make predictions
y_pred = phishing_model.predict(X_test)

# Get probabilities for the phishing class (label 0)
y_proba_phishing = phishing_model.predict_proba(X_test)[:, 0]

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1],
        target_names=["Phishing", "Legitimate"]
    )
)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))

print("\nROC-AUC:")
print(round(roc_auc_score(y_test, y_proba_phishing), 4))
```

```python
# Calculate ROC-AUC with phishing treated as the positive class

y_test_phishing = (y_test == 0).astype(int)

roc_auc = roc_auc_score(
    y_test_phishing,
    y_proba_phishing
)

print("Correct ROC-AUC:", round(roc_auc, 4))
```

```python
import re
from urllib.parse import urlparse
import pandas as pd


def extract_url_features(url):
    """
    Extract lexical features directly from a URL.
    These features do not require visiting the website.
    """

    url = str(url).strip()

    # Make URL parseable even when scheme is missing
    parse_url = url

    if not re.match(r"^[a-zA-Z][a-zA-Z0-9+.-]*://", parse_url):
        parse_url = "//" + parse_url

    parsed = urlparse(parse_url)

    hostname = parsed.hostname or ""
    path = parsed.path or ""
    query = parsed.query or ""

    # Basic character counts
    letters = sum(char.isalpha() for char in url)
    digits = sum(char.isdigit() for char in url)

    special_characters = sum(
        not char.isalnum()
        for char in url
    )

    # IP address detection
    is_ip = int(
        bool(
            re.fullmatch(
                r"\d{1,3}(\.\d{1,3}){3}",
                hostname
            )
        )
    )

    # Domain characteristics
    domain_parts = hostname.split(".") if hostname else []

    if len(domain_parts) >= 2:
        subdomain_count = max(0, len(domain_parts) - 2)
        tld_length = len(domain_parts[-1])
    else:
        subdomain_count = 0
        tld_length = 0

    # Obfuscation indicators
    percent_encoded_count = len(
        re.findall(r"%[0-9A-Fa-f]{2}", url)
    )

    has_obfuscation = int(
        percent_encoded_count > 0
        or "@" in url
        or "\\" in url
    )

    # Ratios
    url_length = len(url)

    letter_ratio = (
        letters / url_length
        if url_length > 0 else 0
    )

    digit_ratio = (
        digits / url_length
        if url_length > 0 else 0
    )

    special_ratio = (
        special_characters / url_length
        if url_length > 0 else 0
    )

    return {
        "url_length": url_length,
        "domain_length": len(hostname),
        "is_ip": is_ip,
        "tld_length": tld_length,
        "subdomain_count": subdomain_count,
        "is_https": int(parsed.scheme.lower() == "https"),
        "letter_count": letters,
        "letter_ratio": letter_ratio,
        "digit_count": digits,
        "digit_ratio": digit_ratio,
        "equals_count": url.count("="),
        "question_count": url.count("?"),
        "ampersand_count": url.count("&"),
        "at_count": url.count("@"),
        "hyphen_count": url.count("-"),
        "dot_count": url.count("."),
        "percent_encoded_count": percent_encoded_count,
        "special_char_ratio": special_ratio
    }
```

```python
test_urls = [
    "https://www.google.com",
    "https://example.com/login",
    "http://192.168.1.25/login",
    "https://secure-login-account-verify.example.com/account?user=test"
]

for url in test_urls:
    print("\nURL:")
    print(url)

    features = extract_url_features(url)

    for feature, value in features.items():
        print(f"{feature}: {value}")
```

```python
# Generate the exact features our application will use

url_feature_data = X["URL"].apply(extract_url_features)

X_lexical = pd.DataFrame(
    url_feature_data.tolist(),
    index=X.index
)

print("Generated feature dataset successfully!")
print("Shape:", X_lexical.shape)

print("\nFeatures:")
print(X_lexical.columns.tolist())

print("\nMissing values:")
print(X_lexical.isnull().sum().sum())
```
